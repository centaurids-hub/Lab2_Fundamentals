```python
# Exercise 1: Code Name

target_name = 'CRESENCIO'
vowels = 'A', 'E', 'I', "O", "U"
SEED_NUM = '3'
redacted_name = ""

for letter in target_name:
    # Check if the individual letter is one of the vowels
    if letter in vowels: 
        redacted_name = (redacted_name + SEED_NUM) #filters vowels, vowels to SEED_NUM
    else:
        redacted_name = (redacted_name + letter) #just prints the consonants
print('Redacted Name:', redacted_name) #prints the loop
```

```python
# Exercise 2: Forensic Signature

First_letter = 'CRESENCIO'[0] #Selects the first letter
ASCII = ord(First_letter) #Converts letter to ACII Value, the ord one is from Cell 3
SEED_NUM = 3
LOOPS = 6
Forensic_Signature = ((ASCII * SEED_NUM) - LOOPS) #Given solution/equation, ((67 * 3)-6)
print('Forensic Signature:', Forensic_Signature)
```

```python
# Exercise 3: Security Password

correct_password = 'centaurids' # My password
attempts = 0 # Initial number of attempts
while True:
    pwd = input("Type your password here:") 
    attempts = attempts + 1 # Every wrong attempts is counted
    if pwd == correct_password:
        print("Unlocked.")
        break # Break terminates the infinite loop, from cell 13
    else:
        print("Access Denied.")
print('attempts:', attempts)
```
